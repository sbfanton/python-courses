## Introducción

Support Vector Machine (SVM) consiste en un algoritmo de machine learning supervisado que se utiliza tanto para clasificación como para regresión. Sin embargo, se lo utiliza mayoritariamente para clasificación y en especial para clasificación binaria.

Hay dos conceptos principales que se encuentran asociados a este algoritmo:
* Hiperplano
* Vectores de soporte (support vectors)
* Márgenes

El **hiperplano** no es otra cosa que una línea (en el caso de clasificaciones con dos características o dimensiones), o bien un plano o hiperplano (en el caso de tres características o más) que separa los puntos u observaciones en grupos o clases. Si la distribución de los puntos hace imposible la división de los mismos en grupos mediante un hiperplano, se suelen aplicar funciones matemáticas conocidas como *kernels* para aumentar la dimensionalidad de los datos para conseguir generar el hiperplano.

Los **vectores de soporte** son los puntos más cercanos de cada clase al hiperplano. Si los mismos cambian, la posición del hiperplano también cambia.

Los **márgenes** son las distancias entre los vectores de soporte y el hiperplano. Cuanto mayor sea dicha distancia, mejor se podrán separar claramente las clases por medio del hiperplano y por lo tanto mejor clasificará el algoritmo.

### Ventajas:

* El algoritmo trabaja mejor con datasets pequeños.
* Trabaja más eficientemente cuando hay un margen claro de separación.
* Trabaja bien datos con varias dimensiones.

### Desventajas:

* No es recomendable para datasets con gran cantidad de datos (observaciones) ya que el tiempo de entrenamiento es mayor.
* No es adecuado para datasets con superposición de observaciones.

## Matemática detrás de support vector machine

### Conceptos Básicos

* Hiperplano: En un espacio de características, un hiperplano es una línea (en 2D), un plano (en 3D) o una generalización de estos en dimensiones superiores. Su función es separar diferentes clases de datos.

* Vectores de Soporte: Son los puntos de datos más cercanos al hiperplano. Estos puntos son cruciales porque determinan la posición y la orientación del hiperplano.

### La Fórmula

La fórmula básica para un hiperplano en un espacio de características se puede expresar como:

$w \cdot x + b = 0$

Donde:

* $w$: es un vector de pesos que define la dirección del hiperplano. La dimensión del mismo variará en función de la cantidad de dimensiones.
* $x$: es el vector de características de un punto de datos.
* $b$: es un sesgo (bias) que ajusta la posición del hiperplano.

### Objetivo de SVM

El objetivo de SVM es encontrar los valores de $w$ y $b$ que maximizan el margen entre las clases. Esto se traduce en dos condiciones:

* Para los puntos de una clase (por ejemplo, clase 1 o valores positivos): $w \cdot x_i + b \geq 1$

* Para los puntos de la otra clase (por ejemplo, clase -1 o valores negativos): $w \cdot x_i + b \leq -1$

### Maximización del Margen

El margen se define como la distancia entre el hiperplano y los vectores de soporte. SVM busca maximizar este margen, lo que se puede formular como un problema de optimización:

$min(\quad\frac{1}{2} ||w||^2)$

Sujeto a las restricciones mencionadas anteriormente. Aquí, $||w||^2$ es la norma del vector de pesos, y minimizarlo ayuda a maximizar el margen.

### Resumen

* Hiperplano: Se define por $w \cdot x + b = 0$.
* Objetivo: Encontrar $w$ y $b$ que maximicen el margen entre las clases.
* Optimización: Se minimiza $\frac{1}{2} ||w||^2$ bajo ciertas restricciones.

## Kernels

![SVM](./img/svm11.png)

### ¿Qué es un kernel en SVM?
Un **kernel** es una función matemática que mide la similitud entre dos puntos en un espacio transformado de mayor dimensión, sin necesidad de calcular explícitamente la transformación.

### ¿Por qué usamos kernels?
- En muchos casos, los datos no son separables por una línea recta en su espacio original.
- Los kernels permiten mapear los datos a un espacio de mayor dimensión donde sí pueden ser separables.
- Se utilizan en **SVM** (Support Vector Machines) y otros algoritmos de Machine Learning.

---

### Principales tipos de Kernels

| Kernel | Fórmula | Descripción |
|---------|-------------|-------------|
| **Lineal** | $K(x, y) = x \cdot y$ | No cambia la dimensión, es útil si los datos ya son separables linealmente. |
| **Polinómico** | $K(x, y) = (x \cdot y + c)^d$ | Aumenta la dimensión de forma polinómica, útil para relaciones más complejas. |
| **RBF (Gaussiano)** | $K(x, y) = \exp\left(-\frac{\|x - y\|^2}{2\sigma^2} \right)$ | Aumenta la dimensión de forma infinita, útil cuando no sabemos qué transformación usar. |
| **Sigmoide** | $K(x, y) = \tanh(a x \cdot y + b)$ | Similar a redes neuronales, pero menos usado en SVM. |

---

### Ejemplo Completo con el Kernel Polinómico
Vamos a resolver un ejemplo desde cero.

### Supongamos que tenemos los puntos en 2D:

$x_1 = (1,2), \quad x_2 = (3,4)$

### Usamos el kernel polinómico con la fórmula:

$K(x, y) = (x \cdot y + c)^d$

Elegimos:
- $c = 1$ (sesgo)
- $d = 2$ (grado del polinomio)

---

#### Paso 1: Calcular el producto escalar
El producto escalar entre $x_1$ y $x_2$ es:

$x_1 \cdot x_2 = (1 \times 3) + (2 \times 4) = 3 + 8 = 11$

---

#### Paso 2: Aplicar la fórmula del Kernel
Sustituyendo en la ecuación:

$K(x_1, x_2) = (11 + 1)^2 = 12^2 = 144$

**Interpretación**: El valor **144** representa la **similitud entre los puntos en el nuevo espacio transformado**.

---

#### Paso 3: Construcción de la Matriz de Similitud
Si tuviéramos más puntos, construiríamos la **matriz de similitud** para el SVM.

Ejemplo con tres puntos:

$x_1 = (1,2), \quad x_2 = (3,4), \quad x_3 = (2,1)$

Calculamos los valores del kernel y formamos la **matriz de Gram**:

$K =
\begin{bmatrix}
K(x_1, x_1) & K(x_1, x_2) & K(x_1, x_3) \\
K(x_2, x_1) & K(x_2, x_2) & K(x_2, x_3) \\
K(x_3, x_1) & K(x_3, x_2) & K(x_3, x_3)
\end{bmatrix}$

Esta matriz se usa en el entrenamiento del SVM para encontrar el **hiperplano de separación** en el espacio de mayor dimensión.

---

### Conclusión
- **Los kernels permiten trabajar en espacios de mayor dimensión sin calcularlos explícitamente**.
- **El kernel polinómico** transforma los datos en un espacio donde pueden ser separables.
- **El SVM usa la matriz de similitud** para clasificar los datos en este nuevo espacio.


### ¿Qué sucede con las características cuando usamos un kernel?

Lo que ocurre al aplicar un kernel (como el polinómico o el RBF) es que **los puntos en el espacio original no se transforman explícitamente** a un nuevo espacio con más características. Es decir, no calculamos nuevas características directamente como si estuviéramos haciendo una transformación matemática explícita de los puntos.

### Entonces, ¿cómo funciona esto realmente?
En lugar de expandir explícitamente las características de cada punto, **el kernel simplemente calcula la similitud entre los puntos en el espacio transformado**. El trabajo del kernel es **hacer que los cálculos ocurran en el espacio de mayor dimensión de manera eficiente**, sin tener que calcular ni almacenar directamente las nuevas características.

---

### ¿Qué ocurre cuando tenemos 1000 puntos?

1. **Espacio Original**:  
   Si tienes 1000 puntos, cada uno con $d$ características (por ejemplo, 2 características en un espacio 2D), entonces tienes una matriz de tamaño $1000 \times 2$ en el caso de 2D.

2. **Transformación con Kernel**:  
   Aunque el kernel transforma los datos a un espacio de **mayor dimensión**, no calculamos explícitamente las nuevas características. Los cálculos se hacen usando la **matriz de similitud** (o matriz de Gram) que contiene los resultados de los productos escalares entre los puntos.  
   
   **En lugar de tener 1000 características por cada punto**, lo que tenemos es una **matriz $1000 \times 1000$** de similitudes entre todos los puntos. Esto **no corresponde directamente a un aumento en el número de características** de los puntos, sino a la representación de las relaciones de similitud en el espacio transformado.

---

### Es decir:

1. **No obtenemos más características de cada punto**.  
   - En el caso del kernel polinómico de grado 2, por ejemplo, **no expandimos los puntos a 3 características** explícitamente.  
   - **El kernel calcula la similitud entre puntos** en un espacio de mayor dimensión sin transformarlos directamente.

2. **La matriz de similitud tiene un tamaño $n \times n$**:  
   - Si tienes **1000 puntos**, la **matriz de similitud** tendrá un tamaño $1000 \times 1000$, que representa las similitudes entre todos los puntos.  
   - **El número de características de cada punto no aumenta** explícitamente, pero estamos trabajando con la matriz de similitudes para hacer la clasificación.

---

### Visualización Conceptual

Imagina que tienes 3 puntos en un espacio 2D:

- **Punto 1**: $(1, 2)$  
- **Punto 2**: $(3, 4)$  
- **Punto 3**: $(2, 1)$  

El kernel polinómico de grado 2 transforma estos puntos en un espacio de mayor dimensión. Pero no calculamos explícitamente las nuevas características de cada punto. En cambio, usamos las **similitudes entre los puntos** para crear la matriz de similitud $K$, como ya vimos:

$K =
\begin{bmatrix}
36  & 144  & 25  \\
144 & 676  & 121 \\
25  & 121  & 36
\end{bmatrix}$

Esta matriz de similitud nos permite entrenar el **SVM** sin necesidad de transformar los puntos directamente. El SVM utiliza esta matriz para encontrar el **hiperplano de separación** en el espacio de mayor dimensión.

---

### Resumen

**No se expanden explícitamente las características de cada punto.**  
En vez de eso, el **kernel calcula las similitudes** entre los puntos en el espacio de mayor dimensión.  
**La matriz de similitud** (\( K \)) tiene un tamaño \( n \times n \), pero **no significa que cada punto tenga más características**, sino que se calculan las relaciones entre todos los puntos para entrenar el modelo.

## Función de pérdida y gradiente descendiente

### 1. Función de Pérdida (Loss Function) en SVM

La función de pérdida en SVM mide cuán bien se ajusta el modelo a los datos. La **pérdida hinge** es la más común en SVM y penaliza las predicciones incorrectas con un margen de tolerancia.

#### Fórmula de la Pérdida Hinge

La función de pérdida hinge para un solo ejemplo de entrenamiento es:

$L(y_i, \mathbf{x}_i) = \max(0, 1 - y_i (\mathbf{w} \cdot \mathbf{x}_i + b))$

Donde:
- $y_i$ es la etiqueta del ejemplo de entrenamiento ($+1$ o $-1$).
- $\mathbf{x}_i$ es el vector de características del ejemplo.
- $\mathbf{w}$ es el vector de pesos.
- $b$ es el sesgo (bias).
- $\mathbf{w} \cdot \mathbf{x}_i$ es el producto punto.

**Interpretación**:
- Si el ejemplo está correctamente clasificado y fuera del margen $( y_i (\mathbf{w} \cdot \mathbf{x}_i + b) > 1$)), la pérdida es 0.
- Si el ejemplo está incorrectamente clasificado o dentro del margen ($y_i (\mathbf{w} \cdot \mathbf{x}_i + b) \leq 1$), la pérdida es positiva.

### 2. Función Objetivo en SVM

El objetivo es **maximizar el margen** entre las clases y minimizar la pérdida:

$\min_{\mathbf{w}, b} \frac{1}{2} \|\mathbf{w}\|^2 + C \sum_{i=1}^N \max(0, 1 - y_i (\mathbf{w} \cdot \mathbf{x}_i + b))$

Donde:
- $\frac{1}{2} \|\mathbf{w}\|^2$ maximiza el margen.
- $C$ controla el equilibrio entre margen y error.
- La suma representa la pérdida hinge de todos los ejemplos.

### 3. Gradiente Descendente en SVM

El **gradiente descendente** se usa para minimizar la función objetivo.

#### Cálculo del Gradiente

1. **Gradiente respecto a** $\mathbf{w}$:

$\nabla_{\mathbf{w}} J(\mathbf{w}, b) = \mathbf{w} + C \sum_{i=1}^N \left( -y_i \mathbf{x}_i \right) \cdot \mathbb{1}_{\left( 1 - y_i (\mathbf{w} \cdot \mathbf{x}_i + b) > 0 \right)}$

Tener en cuenta que la última parte con el 1 subíndice una condición, es indicativo de que la derivada sería esa solo si se cumple la condición que se encuentra en el subíndice.

2. **Gradiente respecto a** $b$:

$\nabla_b J(\mathbf{w}, b) = C \sum_{i=1}^N \left( -y_i \right) \cdot \mathbb{1}_{\left( 1 - y_i (\mathbf{w} \cdot \mathbf{x}_i + b) > 0 \right)}$

### 4. Actualización de Parámetros

En cada iteración del gradiente descendente:

$\mathbf{w} \leftarrow \mathbf{w} - \eta \nabla_{\mathbf{w}} J(\mathbf{w}, b)$

$b \leftarrow b - \eta \nabla_b J(\mathbf{w}, b)$

Donde:
- $\eta$ es la tasa de aprendizaje.

### 5. Ejemplo Práctico

1. Inicializamos $\mathbf{w}$ y $b$ con valores pequeños o aleatorios.
2. Calculamos la pérdida hinge para cada punto de datos.
3. Calculamos el gradiente respecto a $\mathbf{w}$ y $b$.
4. Actualizamos los parámetros usando gradiente descendente.
5. Repetimos hasta la convergencia o alcanzar un número máximo de iteraciones.

### Resumen

- **Función de pérdida hinge**: Penaliza predicciones incorrectas o dentro del margen.
- **Función objetivo**: Maximiza el margen y minimiza la pérdida.
- **Gradiente descendente**: Encuentra los valores óptimos de $\mathbf{w}$ y $b$.

## Ejemplo en Python

En resumen, tenemos entonces la fórmula del hiperplano que separa dos grupos o clases en nuestras observaciones y que se usará para predecir nuevas:

$y = w \dot x + b$

Para ello, para encontrar los mejores valores de $w$ y $b$ que produzcan los menores errores se usará la fórmula del gradiente descendiente, con las derivadas respectivas de la función de pérdida. Siendo el gradiente descendiente:

* Para $w$:

$w = w - \alpha * dw$ 

siendo $dw$ la derivada de la función de pérdida respecto a $w$ evaluada en los valores actuales de $w$ y $b$


* Para $b$:

$b = b - \alpha * db$ 

siendo $db$ la derivada de la función de pérdida respecto a $b$ evaluada en los valores actuales de $w$ y $b$


y siendo $\alpha$ la tasa de aprendizaje, es decir, el tamaño de cada "paso" que se dará en cada itearación para acercar cada vez más a $w$ y $b$ a sus valores óptimos

---

Tener en cuenta que en el siguiente ejemplo en Python, las fórmulas de las derivadas varían un poco respecto a lo anteriormente visto, ya que es en base al video tutorial. Sin embargo, se debería llegar a la misma conclusión.

Las fórmulas en las que se basa para las derivadas son estas:

## **1. Diferencia en la Función de Costo**

Existen distintas variantes de la función de costo en SVM. La que hemos visto antes es:

$J(\mathbf{w}, b) = \frac{1}{2} \|\mathbf{w}\|^2 + C \sum_{i=1}^{N} \max(0, 1 - y_i (\mathbf{w} \cdot \mathbf{x}_i + b))$

Pero en la imagen que muestras, parece que se está utilizando otra versión que incluye un factor de regularización diferente:

$J(\mathbf{w}, b) = \lambda \|\mathbf{w}\|^2 + \sum_{i=1}^{N} \max(0, 1 - y_i (\mathbf{w} \cdot \mathbf{x}_i + b))$

Aquí, en lugar de dividir la norma de \( \mathbf{w} \) entre 2, se usa un coeficiente \( \lambda \). Este cambio afecta las derivadas.

---

## **2. Derivada de la Regularización**

La parte de regularización en la imagen es:

$\lambda \|\mathbf{w}\|^2$

Su derivada respecto a \( \mathbf{w} \) es:

$\frac{\partial}{\partial \mathbf{w}} \lambda \|\mathbf{w}\|^2 = 2\lambda \mathbf{w}$

En la versión que vimos antes, el coeficiente era $\frac{1}{2}$, por lo que la derivada resultaba simplemente en $\mathbf{w}$. Aquí, como el coeficiente es $\lambda$, el resultado es $2\lambda \mathbf{w}$.

---

## **3. Derivada de la Pérdida Hinge**

La pérdida hinge es:

$\sum_{i=1}^{N} \max(0, 1 - y_i (\mathbf{w} \cdot \mathbf{x}_i + b))$

Tomemos un solo dato $(x_i, y_i)$ para analizar la derivada:

- **Caso 1:** Si $y_i (\mathbf{w} \cdot \mathbf{x}_i + b) \geq 1$, la pérdida es 0 y su derivada es **0**.
- **Caso 2:** Si $y_i (\mathbf{w} \cdot \mathbf{x}_i + b) < 1$, la pérdida es $1 - y_i (\mathbf{w} \cdot \mathbf{x}_i + b)$, y su derivada es:

$\frac{\partial}{\partial \mathbf{w}} (1 - y_i (\mathbf{w} \cdot \mathbf{x}_i + b)) = -y_i \mathbf{x}_i$

$\frac{\partial}{\partial b} (1 - y_i (\mathbf{w} \cdot \mathbf{x}_i + b)) = -y_i$

---

## **4. Explicación de las Derivadas en la Imagen**

Ahora entendemos por qué en la imagen se presentan así:

### **Cuando $y_i (\mathbf{w} \cdot \mathbf{x}_i - b) \geq 1$**
- El dato está bien clasificado y fuera del margen.
- Solo contribuye la regularización, por lo que:

$\frac{dJ}{dw} = 2\lambda w, \quad \frac{dJ}{db} = 0$

### **Cuando $y_i (\mathbf{w} \cdot \mathbf{x}_i - b) < 1$**
- El dato está dentro del margen o mal clasificado.
- Se suma la contribución de la pérdida hinge:

$\frac{dJ}{dw} = 2\lambda w - y_i x_i, \quad \frac{dJ}{db} = y_i$

---

## **5. Conclusión**

La diferencia en las derivadas parciales se debe a:
1. **Diferente formulación de la regularización** (factor $\lambda$ en lugar de $\frac{1}{2}$).
2. **Diferente enfoque en la función de costo**, pero el principio general de la pérdida hinge sigue siendo el mismo.

Ambas formulaciones son válidas y se pueden usar dependiendo de la implementación del SVM.

**IMPORTANTE**

Tener en cuenta que la siguiente clase a continuación es una implementación básica de cero del modelo de SVM.
Esta implementación solo funciona para problemas *linealmente separables*. Si los datos no son separables de manera lineal, se necesitaría introducir kernels o usar `sklearn.svm.SVC` con un kernel no lineal.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
```

```python
class SVM_classifier:

    # initiating the hyperparameters
    def __init__(self, learning_rate=0.001, iterations_number=1000, lambda_param=0.01):
        self.learning_rate = learning_rate
        self.iterations_number = iterations_number
        self.lambda_param = lambda_param

    # fitting the model and dataset to SVM classifier
    def fit(self, X, y):
        # m --> filas
        # n --> columnas o features
        self.m, self.n = X.shape

        # initiating weigth and bias values
        self.w = np.zeros(self.n)
        self.b = 0

        self.X = X
        self.y = y

        # implementing gradiente descent algorithm
        for i in range(self.iterations_number):
            self.__update_weights()

    
    # function for updating the weigth and bias values
    def __update_weights(self):
        # label encoding
        y_label = np.where(self.y == 0, -1, 1)

        for index, x_i in enumerate(self.X):
            '''
            La siguiente condicion es para verificar que si el resultado de y_i * (m * x_i + b) es mayor o igual a 1,
            entonces se utilizarían unas derivadas especificas de w y b.
            De lo contrario, se utilizarian otras
            '''
            condition = y_label[index] * (np.dot(x_i, self.w) - self.b) >= 1
    
            if condition:
                dw = 2 * self.lambda_param * self.w
                db = 0
            else:
                dw = 2 * self.lambda_param * self.w - np.dot(x_i, y_label[index])
                db = y_label[index]

            # descent gradient
            self.w = self.w - self.learning_rate * dw
            self.b = self.b - self.learning_rate * db
    

    # predict the label/class for given input values
    def predict(self, X):
        output = np.dot(X, self.w) + self.b
        predicted_labels = np.sign(output) # valor negativo lo convierte a -1, valor positivo lo convierte a 1
        y_hat = np.where(predicted_labels <= -1, 0, 1)
        return y_hat
        
```

```python
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.2, random_state=2)

features.shape, X_train.shape, X_test.shape
```

```python
scaler = StandardScaler().fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

#### Entrenemos el modelo

```python
model = SVM_classifier()
```

```python
model.fit(X_train, y_train)
```

```python
# accuracy on training data
X_train_prediction = model.predict(X_train)
trainig_data_accuracy = accuracy_score(y_train, X_train_prediction)

trainig_data_accuracy
```
