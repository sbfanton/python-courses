# Maldición de la Dimensionalidad

La "maldición de la dimensionalidad" es un término que se utiliza en el campo de la estadística, el aprendizaje automático y la teoría de la información para describir varios fenómenos que surgen cuando se trabaja con datos en espacios de alta dimensión. A medida que el número de dimensiones (o características/columnas) de un conjunto de datos aumenta, varios problemas pueden surgir:

* **Escasez de datos**: En espacios de alta dimensión, los datos tienden a volverse escasos. Esto significa que, para obtener una buena estimación de las propiedades de la población, se necesita una cantidad exponencialmente mayor de datos. Con pocos datos, es difícil generalizar y hacer predicciones precisas.

* **Distancias**: En alta dimensión, las distancias entre puntos se vuelven menos informativas. Por ejemplo, la diferencia entre la distancia más corta y la más larga entre puntos puede ser mínima, lo que dificulta la identificación de patrones y la clasificación.

* **Sobrefitting**: Con un gran número de dimensiones, los modelos pueden ajustarse demasiado a los datos de entrenamiento, capturando ruido en lugar de patrones reales. Esto resulta en un rendimiento deficiente en datos no vistos.

* **Complejidad computacional**: A medida que se añaden más dimensiones, la complejidad computacional de los algoritmos de aprendizaje automático puede aumentar significativamente, lo que puede hacer que el procesamiento de datos sea ineficiente.

Para mitigar la maldición de la dimensionalidad, se pueden utilizar técnicas como la reducción de dimensionalidad (por ejemplo, PCA, t-SNE) o la selección de características, que ayudan a simplificar los datos y a mejorar la eficiencia y la efectividad de los modelos.

## Generar datos aleatorios sintéticos de 3 variables

```python
import numpy as np
```

```python
x = np.random.randint(10, size=10)
y = np.random.randint(10, size=10)
z = np.random.randint(10, size=10)

x, y, z
```

## Graficar los datos para cada dimensión

```python
import matplotlib.pyplot as plt
```

```python
# Crear grafica con tres figuras
fig = plt.figure(figsize=(18, 6))

# espacio unidimensional
ax = fig.add_subplot(1, 3, 1)
ax.plot(x, [0]*10, 
        linewidth=0, marker="*", color="purple", markersize=20)

# espacio bidimensional
ax = fig.add_subplot(1, 3, 2)
ax.scatter(x, y, 
           marker="*", c="purple", s=200)


# espacio tridimensional
ax = fig.add_subplot(1, 3, 3, projection="3d")
ax.scatter(x, y, z, 
           marker="*", c="purple", s=200)

plt.show()
```

## El número de configuraciones se incrementa

```python
fig = plt.figure(figsize=(18, 6))

# Espacio unidimensional
ax = fig.add_subplot(1, 3, 1)

ax.set_title("10^1 = 10", fontsize=35)

ax.plot(range(0, 10), [0]*10, 
        marker="o", linewidth=0,
        color="skyblue", markersize=20)

ax.plot(x, [0]*10, linewidth=0,
        marker="*", color="purple",
        markersize=20)


# Espacio bidimensional
ax = fig.add_subplot(1, 3, 2)

ax.set_title("10^2 = 100", fontsize=35)

for i in range(10):
    ax.scatter(range(0, 10), [i]*10,
               s=200, c="skyblue", marker="*")
    
ax.scatter(x, y, marker="*", c="purple", s=200)


# Espacio tridimensional
ax = fig.add_subplot(1, 3, 3, projection="3d")

ax.set_title("10^3 = 1000", fontsize=35)

for i in range(10):
    for j in range(10):
         ax.scatter(range(0, 10), [i]*10, j, 
                    s=200, c="skyblue", marker="*",
                    alpha=0.25)

ax.scatter(x, y, z, marker="*", c="purple", s=200)

plt.show()
```

En los gráficos anteriores se ha visto claramente cuánto puede afectar la existencia de ua mayor cantidad de dimensiones, características o variables a la hora de realizar análisis de datos para modelos de machine learning.

En estos ejemplos en particular tenemos 10 elevado al número de variables para cada caso. Y por qué 10? Porque aquí los valores que toman las características pueden ir del 0 al 10. Entonces, vemos que cuanto más variables haya, más configuraciones o "puntos" tenemos en dicho espacio de características, y necesitaremos más cantidad de datos que "ocupen" dichos puntos para tener suficiente información.
