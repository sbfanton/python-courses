## Introducción

Support Vector Machine (SVM) consiste en un algoritmo de machine learning supervisado que se utiliza tanto para clasificación como para regresión. Sin embargo, se lo utiliza mayoritariamente para clasificación y en especial para clasificación binaria.

Hay dos conceptos principales que se encuentran asociados a este algoritmo:
* Hiperplano
* Vectores de soporte (support vectors)
* Márgenes

El **hiperplano** no es otra cosa que una línea (en el caso de clasificaciones con dos características o dimensiones), o bien un plano o hiperplano (en el caso de tres características o más) que separa los puntos u observaciones en grupos o clases. Si la distribución de los puntos hace imposible la división de los mismos en grupos mediante un hiperplano, se suelen aplicar funciones matemáticas conocidas como *kernels* para aumentar la dimensionalidad de los datos para conseguir generar el hiperplano.

Los **vectores de soporte** son los puntos más cercanos al hiperplano de cada clase. Si los mismos cambian, la posición del hiperplano también cambia.

Los **márgenes** son las distancias entre los vectores de soporte y el hiperplano. Cuanto mayor sea dicha distancia, mejor se podrán separar claramente las clases por medio del hiperplano y por lo tanto mejor clasificará el algoritmo.

### Ventajas:

* El algoritmo trabaja mejor con datasets pequeños.
* Trabaja más eficientemente cuando hay un margen claro de separación.
* Trabaja bien datos con varias dimensiones.

### Desventajas:

* No es recomendable para datasets con gran cantidad de datos (observaciones) ya que el tiempo de entrenamiento es mayor.
* No es adecuado para datasets con superposición de observaciones.

## Matemática detrás de support vector machine

### Conceptos Básicos

* Hiperplano: En un espacio de características, un hiperplano es una línea (en 2D), un plano (en 3D) o una generalización de estos en dimensiones superiores. Su función es separar diferentes clases de datos.

* Vectores de Soporte: Son los puntos de datos más cercanos al hiperplano. Estos puntos son cruciales porque determinan la posición y la orientación del hiperplano.

### La Fórmula

La fórmula básica para un hiperplano en un espacio de características se puede expresar como:

$w \cdot x + b = 0$

Donde:

* $w$: es un vector de pesos que define la dirección del hiperplano. La dimensión del mismo variará en función de la cantidad de dimensiones.
* $x$: es el vector de características de un punto de datos.
* $b$: es un sesgo (bias) que ajusta la posición del hiperplano.

### Objetivo de SVM

El objetivo de SVM es encontrar los valores de $w$ y $b$ que maximizan el margen entre las clases. Esto se traduce en dos condiciones:

* Para los puntos de una clase (por ejemplo, clase 1 o valores positivos): $w \cdot x_i + b \geq 1$

* Para los puntos de la otra clase (por ejemplo, clase -1 o valores negativos): $w \cdot x_i + b \leq -1$

### Maximización del Margen

El margen se define como la distancia entre el hiperplano y los vectores de soporte. SVM busca maximizar este margen, lo que se puede formular como un problema de optimización:

$min(\quad\frac{1}{2} ||w||^2)$

Sujeto a las restricciones mencionadas anteriormente. Aquí, $||w||^2$ es la norma del vector de pesos, y minimizarlo ayuda a maximizar el margen.

### Resumen

* Hiperplano: Se define por $w \cdot x + b = 0$.
* Objetivo: Encontrar $w$ y $b$ que maximicen el margen entre las clases.
* Optimización: Se minimiza $\frac{1}{2} ||w||^2$ bajo ciertas restricciones.

## Kernels

![SVM](./img/svm11.png)

### ¿Qué es un kernel en SVM?
Un **kernel** es una función matemática que mide la similitud entre dos puntos en un espacio transformado de mayor dimensión, sin necesidad de calcular explícitamente la transformación.

### ¿Por qué usamos kernels?
- En muchos casos, los datos no son separables por una línea recta en su espacio original.
- Los kernels permiten mapear los datos a un espacio de mayor dimensión donde sí pueden ser separables.
- Se utilizan en **SVM** (Support Vector Machines) y otros algoritmos de Machine Learning.

---

### Principales tipos de Kernels

| Kernel | Fórmula | Descripción |
|---------|-------------|-------------|
| **Lineal** | $K(x, y) = x \cdot y$ | No cambia la dimensión, es útil si los datos ya son separables linealmente. |
| **Polinómico** | $K(x, y) = (x \cdot y + c)^d$ | Aumenta la dimensión de forma polinómica, útil para relaciones más complejas. |
| **RBF (Gaussiano)** | $K(x, y) = \exp\left(-\frac{\|x - y\|^2}{2\sigma^2} \right)$ | Aumenta la dimensión de forma infinita, útil cuando no sabemos qué transformación usar. |
| **Sigmoide** | $K(x, y) = \tanh(a x \cdot y + b)$ | Similar a redes neuronales, pero menos usado en SVM. |

---

### Ejemplo Completo con el Kernel Polinómico
Vamos a resolver un ejemplo desde cero.

### Supongamos que tenemos los puntos en 2D:

$x_1 = (1,2), \quad x_2 = (3,4)$

### Usamos el kernel polinómico con la fórmula:

$K(x, y) = (x \cdot y + c)^d$

Elegimos:
- $c = 1$ (sesgo)
- $d = 2$ (grado del polinomio)

---

#### Paso 1: Calcular el producto escalar
El producto escalar entre $x_1$ y $x_2$ es:

$x_1 \cdot x_2 = (1 \times 3) + (2 \times 4) = 3 + 8 = 11$

---

#### Paso 2: Aplicar la fórmula del Kernel
Sustituyendo en la ecuación:

$K(x_1, x_2) = (11 + 1)^2 = 12^2 = 144$

**Interpretación**: El valor **144** representa la **similitud entre los puntos en el nuevo espacio transformado**.

---

#### Paso 3: Construcción de la Matriz de Similitud
Si tuviéramos más puntos, construiríamos la **matriz de similitud** para el SVM.

Ejemplo con tres puntos:

$x_1 = (1,2), \quad x_2 = (3,4), \quad x_3 = (2,1)$

Calculamos los valores del kernel y formamos la **matriz de Gram**:

$K =
\begin{bmatrix}
K(x_1, x_1) & K(x_1, x_2) & K(x_1, x_3) \\
K(x_2, x_1) & K(x_2, x_2) & K(x_2, x_3) \\
K(x_3, x_1) & K(x_3, x_2) & K(x_3, x_3)
\end{bmatrix}$

Esta matriz se usa en el entrenamiento del SVM para encontrar el **hiperplano de separación** en el espacio de mayor dimensión.

---

### Conclusión
- **Los kernels permiten trabajar en espacios de mayor dimensión sin calcularlos explícitamente**.
- **El kernel polinómico** transforma los datos en un espacio donde pueden ser separables.
- **El SVM usa la matriz de similitud** para clasificar los datos en este nuevo espacio.


### ¿Qué sucede con las características cuando usamos un kernel?

Lo que ocurre al aplicar un kernel (como el polinómico o el RBF) es que **los puntos en el espacio original no se transforman explícitamente** a un nuevo espacio con más características. Es decir, no calculamos nuevas características directamente como si estuviéramos haciendo una transformación matemática explícita de los puntos.

### Entonces, ¿cómo funciona esto realmente?
En lugar de expandir explícitamente las características de cada punto, **el kernel simplemente calcula la similitud entre los puntos en el espacio transformado**. El trabajo del kernel es **hacer que los cálculos ocurran en el espacio de mayor dimensión de manera eficiente**, sin tener que calcular ni almacenar directamente las nuevas características.

---

### ¿Qué ocurre cuando tenemos 1000 puntos?

1. **Espacio Original**:  
   Si tienes 1000 puntos, cada uno con $d$ características (por ejemplo, 2 características en un espacio 2D), entonces tienes una matriz de tamaño $1000 \times 2$ en el caso de 2D.

2. **Transformación con Kernel**:  
   Aunque el kernel transforma los datos a un espacio de **mayor dimensión**, no calculamos explícitamente las nuevas características. Los cálculos se hacen usando la **matriz de similitud** (o matriz de Gram) que contiene los resultados de los productos escalares entre los puntos.  
   
   **En lugar de tener 1000 características por cada punto**, lo que tenemos es una **matriz $1000 \times 1000$** de similitudes entre todos los puntos. Esto **no corresponde directamente a un aumento en el número de características** de los puntos, sino a la representación de las relaciones de similitud en el espacio transformado.

---

### Es decir:

1. **No obtenemos más características de cada punto**.  
   - En el caso del kernel polinómico de grado 2, por ejemplo, **no expandimos los puntos a 3 características** explícitamente.  
   - **El kernel calcula la similitud entre puntos** en un espacio de mayor dimensión sin transformarlos directamente.

2. **La matriz de similitud tiene un tamaño $n \times n$**:  
   - Si tienes **1000 puntos**, la **matriz de similitud** tendrá un tamaño $1000 \times 1000$, que representa las similitudes entre todos los puntos.  
   - **El número de características de cada punto no aumenta** explícitamente, pero estamos trabajando con la matriz de similitudes para hacer la clasificación.

---

### Visualización Conceptual

Imagina que tienes 3 puntos en un espacio 2D:

- **Punto 1**: $(1, 2)$  
- **Punto 2**: $(3, 4)$  
- **Punto 3**: $(2, 1)$  

El kernel polinómico de grado 2 transforma estos puntos en un espacio de mayor dimensión. Pero no calculamos explícitamente las nuevas características de cada punto. En cambio, usamos las **similitudes entre los puntos** para crear la matriz de similitud $K$, como ya vimos:

$K =
\begin{bmatrix}
36  & 144  & 25  \\
144 & 676  & 121 \\
25  & 121  & 36
\end{bmatrix}$

Esta matriz de similitud nos permite entrenar el **SVM** sin necesidad de transformar los puntos directamente. El SVM utiliza esta matriz para encontrar el **hiperplano de separación** en el espacio de mayor dimensión.

---

### Resumen

**No se expanden explícitamente las características de cada punto.**  
En vez de eso, el **kernel calcula las similitudes** entre los puntos en el espacio de mayor dimensión.  
**La matriz de similitud** (\( K \)) tiene un tamaño \( n \times n \), pero **no significa que cada punto tenga más características**, sino que se calculan las relaciones entre todos los puntos para entrenar el modelo.
