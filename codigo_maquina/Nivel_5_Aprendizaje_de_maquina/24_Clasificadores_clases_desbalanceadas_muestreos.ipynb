# Clasificadores en clases desbalanceadas usando submuestreo y sobremuestreo

## Submuestreo (Undersampling) y Sobremuestreo (Oversampling) en Problemas de Clases Desbalanceadas

El **submuestreo** y el **sobremuestreo** son dos técnicas comunes para manejar el **desbalanceo de clases** en problemas de clasificación. Ambas se utilizan para equilibrar el número de ejemplos entre las clases mayoritaria y minoritaria.

### 1. Submuestreo (Undersampling)
El submuestreo consiste en reducir el número de muestras de la clase mayoritaria para equilibrar las clases. Esto puede ser útil cuando tienes un número muy grande de datos de una clase y no quieres que esa clase domine el modelo.

#### Ejemplo:
Supón que tienes un conjunto de datos para predecir si un correo es **spam** o **no spam**, y tienes 1000 correos en total:
- 950 correos **no spam**
- 50 correos **spam**

El submuestreo eliminaría algunos de los correos **no spam** para igualar la cantidad de muestras entre ambas clases.

#### Proceso:
- Reducirías los 950 correos **no spam** a solo 50, igualando así el número de muestras en ambas clases: 50 **spam** y 50 **no spam**.

#### Ventajas:
- Puede ser útil si tienes un conjunto de datos muy grande y quieres reducir el tiempo de entrenamiento.
- A veces, puede mejorar el rendimiento si las clases desbalanceadas están muy sesgadas.

#### Desventajas:
- Perderías información importante de la clase mayoritaria, lo que podría disminuir la capacidad del modelo para aprender patrones completos de esa clase.

### 2. Sobremuestreo (Oversampling)
El sobremuestreo implica aumentar el número de muestras de la clase minoritaria, generalmente duplicando ejemplos o generando nuevos ejemplos artificialmente (por ejemplo, usando **SMOTE**).

#### Ejemplo:
En el mismo caso de correos **spam** y **no spam**:
- 950 correos **no spam**
- 50 correos **spam**

El sobremuestreo aumentaría el número de ejemplos de la clase **spam** para igualarlos a los correos **no spam**.

#### Proceso:
- Generarías 900 copias adicionales de los correos **spam**, para tener en total 950 ejemplos de **spam** y 950 de **no spam**.

#### Ventajas:
- No se pierde información de la clase minoritaria.
- Puede ayudar a que el modelo aprenda mejor sobre la clase minoritaria.

#### Desventajas:
- Puede causar sobreajuste (overfitting) porque el modelo podría aprender patrones específicos de las copias generadas, en lugar de patrones generales.
- Aumenta el tamaño del conjunto de datos, lo que puede incrementar el tiempo de entrenamiento.

### Resumen Visual:

Supón que tienes el siguiente conjunto de datos (antes de aplicar submuestreo o sobremuestreo):

| Clase      | Cantidad |  
|------------|----------|  
| **No Spam** | 950      |  
| **Spam**    | 50       |  

- **Submuestreo:** Reducirías los **No Spam** a 50, y tendrías 50 **Spam** y 50 **No Spam**.
  
- **Sobremuestreo:** Aumentarías los **Spam** a 950 (puedes duplicarlos o generar datos sintéticos), y tendrías 950 **Spam** y 950 **No Spam**.

### ¿Qué técnica usar?
- **Submuestreo** es útil si tienes muchos datos y deseas reducir el tiempo de entrenamiento sin perder información crítica.
- **Sobremuestreo** es útil cuando la clase minoritaria tiene pocos datos, y quieres preservar toda la información posible de esa clase.

Ambas técnicas pueden ser combinadas con otras estrategias como **ajustes de pesos de clase** para obtener un modelo robusto.


## Ejemplo en Python

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score 
```

![Desbalanceo 1](./img/desbalanceo1.png)

![Desbalanceo 2](./img/desbalanceo2.png)

```python
# Código de soporte 

def matriz_de_confusion(clases_reales, clases_predichas, titulo, cols):
    """ Visualiza la matriz de confusión """
    print("Reales:", clases_reales.reshape(-1))
    print("Predichas:", clases_predichas)
    matriz = confusion_matrix(clases_reales, clases_predichas)
    accuracy = accuracy_score(clases_reales, clases_predichas)
    
    #Código de matplotlib para graficar 
    plt.figure(figsize=(4, 4))
    matriz = pd.DataFrame(matriz, 
                          columns=cols)
    plt.matshow(matriz, cmap="Pastel1", vmin=0, vmax=20, fignum=1)
    plt.xticks(range(len(matriz.columns)), matriz.columns, rotation=45)
    plt.yticks(range(len(matriz.columns)), matriz.columns)
    etiquetas = (("Verdaderos\nnegativos", "Falsos\npositivos"),
                 ("Falsos\nnegativos", "Verdaderos\npositivos"))
    plt.text(1.60, -0.30, titulo, fontsize=25, c="red")
    plt.text(2.1, 0.10, "Accuracy: %0.2f" % accuracy, fontsize=20)
    for i in range(len(matriz.columns)):
        for j in range(len(matriz.columns)):
            plt.text(i, j + 0.14, str(matriz.iloc[i, j]),
                     fontsize=30, ha="center", va="center")
            plt.text(i, j - 0.25, etiquetas[i][j],
                     fontsize=11.5, ha="center", va="center")           
    plt.show()
```

### Distribución de clases desbalanceadas

```python
# University of Wisconsin, Clinical Sciences Center 
# 30 características: radio, textura, area, entre otros
# 0 : SANAS, 1 : CÁNCER
personas = pd.read_csv("../datos/cancer_desbalance.csv", header=None)
prueba = pd.read_csv("../datos/cancer_prueba.csv", header=None)

# Gráficas de barras
num_sanas = personas[personas[30]==0][0].size
num_cancer = personas[personas[30]==1][0].size
plt.bar(["Sanas (%d)" % num_sanas, "Cáncer (%d)" % num_cancer], 
        [num_sanas, num_cancer],
        color =["cyan", "red"],
        width = 0.8)
plt.ylabel("Personas")
plt.show()
```

### Separando los datos por clase

```python
datos_cancer = personas[personas[30]==1]
datos_sanas = personas[personas[30]==0]
```

### Sobremuestreo con reemplazo

```python
# Generamos misma cantidad de instancias con cancer que la cantidad de instancias de personas sanas
sobremuestreo_cancer = datos_cancer.sample(n=290, replace=True, random_state=0)
sobremuestreo_cancer
```

### Submuestreo sin reemplazo

```python
# Toammso una submuestra de las instancias de la clase mayoritaria para igualar a la cantidad de instancias de la clase minoritaria
submuestreo_sanas = datos_sanas.sample(n=20, replace=False, random_state=0)
submuestreo_sanas
```

### Preparando los datos para las tres clasificaciones

```python
sobremuestro = pd.concat([sobremuestreo_cancer, datos_sanas])
datos_sobremuestreo = sobremuestro.iloc[:, :-1]
clase_sobremuestreo = sobremuestro.iloc[:, -1:]

submuestreo = pd.concat([datos_cancer, submuestreo_sanas])
datos_submuestreo = submuestreo.iloc[:, :-1]
clase_submuestreo = submuestreo.iloc[:, -1:]

desbalanceado = pd.concat([datos_cancer, datos_sanas])
datos_desbalanceado = desbalanceado.iloc[:, :-1]
clase_desbalanceado = desbalanceado.iloc[:, -1:]

# Datos para probar los modelos (son solo 20 instancias)
datos_prueba = prueba.iloc[:, :-1]
clase_prueba = prueba.iloc[:, -1:]
```

### Creación y evaluación de 3 modelos

#### Regresión Logística

```python
modelo = LogisticRegression().fit(datos_desbalanceado.values, 
                                  clase_desbalanceado.values.reshape(-1))
matriz_de_confusion(clase_prueba.values, 
                    modelo.predict(datos_prueba.values), 
                    "Desbalanceado",
                    ["0: Sano", "1: Cáncer"]) 

modelo = LogisticRegression().fit(datos_submuestreo.values, 
                                  clase_submuestreo.values.reshape(-1))
matriz_de_confusion(clase_prueba.values, 
                    modelo.predict(datos_prueba.values), 
                    "Submuestreo",
                   ["0: Sano", "1: Cáncer"]) 

modelo = LogisticRegression().fit(datos_sobremuestreo.values, 
                                  clase_sobremuestreo.values.reshape(-1))
matriz_de_confusion(clase_prueba.values, 
                    modelo.predict(datos_prueba.values), 
                    "Sobremuestreo",
                   ["0: Sano", "1: Cáncer"]) 
```
