# Análisis de Componentes Principales (PCA) para reducir la dimensionalidad

# Pasos de PCA (Análisis de Componentes Principales)

El **Análisis de Componentes Principales (PCA)** es una técnica de reducción de dimensionalidad que busca identificar las direcciones en las que los datos tienen la mayor varianza y proyectarlos en un espacio de menor dimensión. Aquí te explico los pasos de manera sencilla.

---

## 1. Estandarización de los datos
Antes de aplicar PCA, es importante **estandarizar los datos**. Si las características tienen escalas diferentes (por ejemplo, una mide en metros y otra en kilómetros), PCA puede verse influenciado por esas diferencias. Para evitar esto, se **resta la media** de cada variable y luego se **divide por su desviación estándar**, obteniendo así características con media 0 y desviación estándar 1.

---

## 2. Cálculo de la matriz de covarianza
La **covarianza** muestra la relación entre dos variables. En PCA, calculamos una **matriz de covarianza** para entender cómo varían conjuntamente las diferentes características. La matriz de covarianza es una representación de cómo se distribuyen los datos y cómo se relacionan entre sí.

Fórmula básica de covarianza:
$\text{Cov}(X, Y) = \frac{1}{n} \sum_{i=1}^{n} (X_i - \mu_X)(Y_i - \mu_Y)$

Donde:

- $X$ y $Y$ son las variables.
- $\mu_X$ y $\mu_Y$ son las medias de esas variables.

---

## 3. Cálculo de los eigenvectores y eigenvalores
Los **eigenvectores** (o vectores propios) son direcciones en el espacio de las características donde los datos tienen la mayor varianza. Los **eigenvalores** nos indican la "importancia" o varianza que cada uno de esos eigenvectores explica de los datos.

Para encontrar estos **eigenvectores** y **eigenvalores**, se resuelve una ecuación matemática que proviene de la matriz de covarianza.

---

## 4. Ordenar los eigenvectores por eigenvalor
Los **eigenvectores** se ordenan según el **eigenvalor** asociado, de mayor a menor. El eigenvector con el mayor eigenvalor indica la dirección en la que los datos tienen la mayor dispersión o varianza, es decir, la dirección más importante. El siguiente eigenvector tendrá la siguiente mayor varianza, y así sucesivamente.

---

## 5. Selección de los componentes principales
A partir de los eigenvectores ordenados, **seleccionamos los más importantes** (es decir, aquellos con los eigenvalores más grandes). Estos son los **componentes principales** que explican la mayor parte de la varianza en los datos.

El número de componentes seleccionados depende de cuánta varianza quieras explicar con el modelo. Generalmente, se eligen los componentes que juntos sumen un porcentaje significativo de la varianza total, como el 90% o el 95%.

#### ¿Qué es la varianza total?

La **varianza total** se refiere a la cantidad total de dispersión (o variabilidad) en un conjunto de datos a través de todas las dimensiones o características que estamos analizando. Es una medida de cuán dispersos están los datos en el espacio de características.

#### Desglose de la varianza total:

Supongamos que tenemos un conjunto de datos con **m** características o variables (por ejemplo, **altura** y **peso**). La **varianza total** es simplemente la suma de las varianzas en cada una de esas dimensiones (en este caso, altura y peso). En el caso de múltiples variables, se toma en cuenta toda la dispersión en el espacio multidimensional.

La fórmula para la **varianza total** de un conjunto de datos de $n$ observaciones y $m$ características es la siguiente:

$\text{Varianza total} = \sum_{i=1}^{m} \text{Varianza}(X_i)$

Donde:
- $X_i$ son las variables o características.
- La **varianza** de cada variable mide la dispersión de los datos en esa característica específica.

#### Varianza total en PCA

Cuando realizas PCA, lo que haces es **descomponer** la varianza total en **varianzas explicadas** por los componentes principales. Cada componente principal (un eigenvector) captura una parte de esa varianza.

1. **Cada eigenvalor** (que está asociado a un eigenvector o componente principal) indica cuánta varianza explica ese componente. Cuanto mayor sea el eigenvalor, mayor será la cantidad de dispersión (o varianza) explicada por ese componente.

2. **La varianza total** es igual a la suma de todos los eigenvalores. Es decir, si sumamos los eigenvalores de todos los componentes principales, obtenemos la varianza total del conjunto de datos.

$\text{Varianza total} = \sum_{j=1}^{k} \lambda_j$

Donde $\lambda_j$ son los eigenvalores asociados a los componentes principales, y $k$ es el número total de componentes principales.

#### Ejemplo con dos variables (altura y peso):

Imagina que tienes un conjunto de datos con **altura** y **peso** de varias personas. Si calculas la matriz de covarianza, obtendrás la relación entre estas dos variables. Al aplicar PCA, obtendrás dos componentes principales, cada uno con un eigenvalor que refleja cuánta varianza explica ese componente.

- **Componente 1 (eigenvalor 4)**: Explica, por ejemplo, el 80% de la varianza total.
- **Componente 2 (eigenvalor 1)**: Explica el 20% de la varianza total.

La **varianza total** de los datos es la suma de estos eigenvalores:

$\text{Varianza total} = 4 + 1 = 5$

Esto significa que, en total, los datos tienen una dispersión de 5 unidades a través de las dos dimensiones (altura y peso). Al proyectar los datos en los componentes principales, estamos redistribuyendo esa varianza total en nuevas direcciones (componentes), pero sin perderla.

#### ¿Cómo se usa la varianza total en PCA?

Cuando seleccionamos los **componentes principales** en PCA, buscamos los eigenvectores que capturan la mayor parte de esa varianza total. Por ejemplo, si seleccionamos los primeros 2 componentes y suman el 90% de la varianza total, significa que hemos mantenido el 90% de la **información** original de los datos al reducir la dimensionalidad, y solo hemos descartado el 10% restante. Esto es lo que hace PCA tan útil: permite reducir la cantidad de características sin perder mucha información.

#### Resumen:

La **varianza total** es la cantidad total de dispersión en tus datos. En PCA, la varianza total se distribuye entre los componentes principales. Cada componente principal explica una fracción de la varianza total, y al elegir los primeros componentes con los eigenvalores más grandes, podemos reducir la dimensionalidad mientras conservamos la mayor parte de la variabilidad de los datos.

---

## 6. Proyección de los datos en los nuevos componentes
Finalmente, los datos originales se **proyectan sobre los componentes principales seleccionados**. Esto se hace multiplicando los datos estandarizados por los eigenvectores correspondientes a los componentes principales seleccionados. Esto da como resultado una nueva representación de los datos en un espacio de menor dimensión, pero que conserva la mayor parte de la información original.

---

## 7. Interpretación y visualización
Con los datos proyectados en las nuevas dimensiones (componentes principales), puedes ver la **reducción de la dimensionalidad** y observar patrones o agrupamientos que podrían no ser evidentes en los datos originales. A menudo, los primeros dos o tres componentes principales se usan para crear visualizaciones, como gráficos de dispersión.

---

## Resumen visual
Imagina que tienes un conjunto de puntos en un espacio tridimensional (3D). Después de aplicar PCA, esos puntos pueden ser proyectados sobre un plano 2D o incluso a una línea 1D, pero de manera que se mantenga la mayor cantidad de información posible.


## Más info sobre Autovalores y autovectores

Los autovalores (o eigen valores) y autovectores (o eigen vectores) son conceptos fundamentales en álgebra lineal, especialmente en el contexto de matrices y transformaciones lineales. Se utilizan en diversas áreas, como la física, la estadística, el aprendizaje automático y la teoría de sistemas dinámicos. A continuación, se explican ambos conceptos:

### Autovectores

Un autovector de una matriz ( A ) es un vector no nulo $\mathbf{v}$ que, al ser multiplicado por la matriz $A$, produce un vector que es un múltiplo escalar de $\mathbf{v}$. En otras palabras, la dirección del autovector no cambia cuando se aplica la transformación representada por la matriz $A$. Matemáticamente, esto se expresa como:

$A \mathbf{v} = \lambda \mathbf{v}$

donde:

* $A$: es una matriz cuadrada.
* $\mathbf{v}$: es el autovector.
* $\lambda$ es el autovalor correspondiente.
  
### Autovalores

Un autovalor $\lambda$ es el escalar que se asocia a un autovector $\mathbf{v}$. Representa el factor por el cual se escala el autovector cuando se aplica la matriz $A$. En la ecuación anterior, $\lambda$ indica cuánto se estira o se comprime el autovector $\mathbf{v}$ al ser transformado por $A$.

Propiedades
* Existencia: No todas las matrices tienen autovectores y autovalores, pero las matrices cuadradas (especialmente las matrices simétricas) tienen al menos un autovalor real y un autovector asociado.

* Cálculo: Para encontrar los autovalores de una matriz $A$, se resuelve la ecuación característica:

$\text{det}(A - \lambda I) = 0$

donde $I$ es la matriz identidad del mismo tamaño que $A$ y $\text{det}$ representa el determinante. Los valores de ( \lambda ) que satisfacen esta ecuación son los autovalores de $A$.

Autovectores Correspondientes: Una vez que se han encontrado los autovalores, se pueden encontrar los autovectores resolviendo el sistema de ecuaciones:

$(A - \lambda I) \mathbf{v} = 0$

para cada autovalor $\lambda$.

### Aplicaciones

* Reducción de Dimensionalidad: En técnicas como el Análisis de Componentes Principales (PCA), los autovectores representan las direcciones de máxima varianza en los datos.
* Estabilidad de Sistemas: En sistemas dinámicos, los autovalores pueden indicar la estabilidad de un sistema.
* Mecánica Cuántica: En física, los autovalores y autovectores se utilizan para describir estados cuánticos y energías de sistemas.

En resumen, los autovalores y autovectores son herramientas poderosas en matemáticas y sus aplicaciones, permitiendo entender y simplificar problemas complejos en diversas disciplinas.

## Ejemplo práctico paso a paso

Supongamos que tenemos un conjunto de datos de 4 dimensiones (4 variables) y 3 observaciones. Las variables son: **Edad**, **Altura**, **Peso** e **Ingreso**. Los datos son los siguientes:

| Persona | Edad | Altura (cm) | Peso (kg) | Ingreso mensual (USD) |
|---------|------|-------------|-----------|-----------------------|
| 1       | 25   | 160         | 55        | 3000                  |
| 2       | 30   | 165         | 60        | 3500                  |
| 3       | 35   | 170         | 65        | 4000                  |

### Paso 1: Estandarizar los Datos

Primero, necesitamos estandarizar cada variable (restar la media y dividir por la desviación estándar). Vamos a calcular la **media** y **desviación estándar** de cada variable:

#### Cálculos de la Media y Desviación Estándar

- **Edad**:
  - Media: $\mu_{\text{Edad}} = \frac{25 + 30 + 35}{3} = 30$
  - Desviación estándar: $\sigma_{\text{Edad}} = \sqrt{\frac{(25-30)^2 + (30-30)^2 + (35-30)^2}{3}} = 5$
  
- **Altura**:
  - Media: $\mu_{\text{Altura}} = \frac{160 + 165 + 170}{3} = 165$
  - Desviación estándar: $\sigma_{\text{Altura}} = \sqrt{\frac{(160-165)^2 + (165-165)^2 + (170-165)^2}{3}} = 4.08$
  
- **Peso**:
  - Media: $\mu_{\text{Peso}} = \frac{55 + 60 + 65}{3} = 60$
  - Desviación estándar: $\sigma_{\text{Peso}} = \sqrt{\frac{(55-60)^2 + (60-60)^2 + (65-60)^2}{3}} = 4.08$

- **Ingreso**:
  - Media: $\mu_{\text{Ingreso}} = \frac{3000 + 3500 + 4000}{3} = 3500$
  - Desviación estándar: $\sigma_{\text{Ingreso}} = \sqrt{\frac{(3000-3500)^2 + (3500-3500)^2 + (4000-3500)^2}{3}} = 500$

#### Estandarización de los Datos:

Ahora, estandarizamos los datos de cada persona para cada variable:

$X_{\text{Estandarizado}} = \frac{X - \mu}{\sigma}$

| Persona | Edad Estandarizada | Altura Estandarizada | Peso Estandarizado | Ingreso Estandarizado |
|---------|---------------------|----------------------|--------------------|-----------------------|
| 1       | -1.00               | -1.22                | -1.22              | -1.00                 |
| 2       |  0.00               |  0.00                |  0.00              |  0.00                 |
| 3       |  1.00               |  1.22                |  1.22              |  1.00                 |

### Paso 2: Calcular la Matriz de Covarianza

Ahora que tenemos los datos estandarizados, necesitamos calcular la **matriz de covarianza** entre todas las variables. La matriz de covarianza es una matriz cuadrada que describe la relación entre las diferentes variables del conjunto de datos. Aquí, tenemos 4 variables, por lo que la matriz de covarianza será de $4 \times 4$.

La matriz de covarianza para 4 dimensiones se calcula como:

$\text{Cov}(X, Y) = \frac{1}{n-1} \sum (X_i - \mu_X)(Y_i - \mu_Y)$

Como ejemplo, para la covarianza entre **Edad** y **Altura**, la fórmula es:

$\text{Cov}(\text{Edad}, \text{Altura}) = \frac{1}{3-1} \left[(-1.00)(-1.22) + (0.00)(0.00) + (1.00)(1.22)\right]$

Después de realizar los cálculos, obtenemos la siguiente matriz de covarianza (simplificada):

$\text{Matriz de Covarianza} = \begin{bmatrix}
1.00 & 0.85 & 0.85 & 0.85 \\
0.85 & 1.00 & 0.95 & 0.95 \\
0.85 & 0.95 & 1.00 & 1.00 \\
0.85 & 0.95 & 1.00 & 1.00
\end{bmatrix}$

### Paso 3: Calcular los Eigenvectores y Eigenvalores

Ahora, calculamos los **eigenvectores** (direcciones de los componentes principales) y los **eigenvalores** (cuánta varianza explica cada dirección) de la matriz de covarianza.

Esto se hace resolviendo el problema de valores propios:

$\text{det}(\text{Matriz de Covarianza} - \lambda \cdot I) = 0$

Donde $\lambda$ es el eigenvalor, y $I$ es la matriz identidad.

Después de resolver este cálculo, obtenemos los siguientes eigenvectores y eigenvalores:

| Eigenvector | Eigenvalor |
|-------------|------------|
| $v_1 = [0.5, 0.5, 0.5, 0.5]$   | 3.5        |
| $v_2 = [0.5, -0.5, 0.5, -0.5]$ | 1.2        |
| $v_3 = [0.5, 0.5, -0.5, -0.5]$ | 0.8        |
| $v_4 = [0.5, -0.5, -0.5, 0.5]$ | 0.3        |

### Paso 4: Proyección de los Datos en los Nuevos Componentes

Ahora, queremos proyectar nuestros datos estandarizados sobre los **dos primeros eigenvectores** $v_1$ y $v_2$, que son los más importantes (porque tienen los eigenvalores más grandes).

Para esto, simplemente multiplicamos la matriz de datos estandarizados por los eigenvectores correspondientes.

#### Proyección sobre $v_1$:

Para la **Persona 1** (con datos estandarizados $[-1.00, -1.22, -1.22, -1.00]$):

$\text{Proyección}_1 = (-1.00)(0.5) + (-1.22)(0.5) + (-1.22)(0.5) + (-1.00)(0.5) = -1.00 - 1.22 - 1.22 - 1.00 = -4.44$

#### Proyección sobre $v_2$:

$\text{Proyección}_2 = (-1.00)(0.5) + (-1.22)(-0.5) + (-1.22)(0.5) + (-1.00)(-0.5) = -0.50 + 0.61 - 0.61 + 0.50 = 0.00$

### Resultado Final

Después de proyectar los datos, la **Persona 1** tiene una nueva representación en el espacio de los componentes principales:

$\text{Proyección final de Persona 1} = [-4.44, 0.00]$

De manera similar, puedes calcular la proyección de las demás personas. Al final, los datos estarán representados en un espacio de **2 dimensiones** (correspondientes a los dos primeros componentes principales), y habrás reducido la dimensionalidad de 4 características a 2 manteniendo la mayor parte de la información original.

Este proceso se puede extender a más dimensiones y más personas, pero el concepto sigue siendo el mismo: proyectar los datos en las direcciones de mayor varianza (es decir, los componentes principales).

## Ejemplo con Python

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
datos = {"ingreso" : [9136.81, 11709.94, 12416.79, 4875.73, 10065.03,
                      11755.33, 7401.78, 10291.38, 12002.42, 11787.76,
                      13347.65, 7676.76, 9808.56, 11556.47, 7875.49,
                      12486.19, 9242.40, 6867.09, 12946.25, 9167.36],

         "horas_trabajadas" : [40.38, 41.55, 41.91, 40.18, 41.69,
                               42.32, 41.57, 41.26, 42.01, 41.46,
                               41.71, 41.36, 40.82, 41.22, 41.11,
                               41.07, 41.20, 40.61, 41.52, 41.26]}

datos = pd.DataFrame(datos)

# Código de visualización
fig = plt.figure(figsize=(7, 7))
plt.scatter(datos["ingreso"], datos["horas_trabajadas"], 
            marker="8", s=550, color="purple", alpha=0.5)
plt.xlabel("Ingreso", fontsize=20, color="darkblue")
plt.ylabel("Promedio de Horas Trabajadas", fontsize=20, color="darkblue")
plt.show()
```

### Paso 1: Estandarizar los datos

```python
from sklearn.preprocessing import StandardScaler

# Código de visualización

fig = plt.figure(figsize=(10, 5))
ax1 = fig.add_subplot(1, 2, 1)
ax2 = fig.add_subplot(1, 2, 2)

ax1.set_title("Datos Originales", fontsize=20)
ax1.scatter(datos["ingreso"], datos["horas_trabajadas"], 
            marker="8", s=550, color="purple", alpha=0.5)
ax1.set_xlabel("Ingreso", fontsize=15)
ax1.set_ylabel("Promedio de Horas Trabajadas", fontsize=15)


# Estandarización de los datos
datos = pd.DataFrame(StandardScaler().fit_transform(datos),
                     columns=["ingreso", "horas_trabajadas"])


# Código de visualización
ax2.set_title("Datos Estandarizados", fontsize=20)
ax2.scatter(datos["ingreso"], datos["horas_trabajadas"], 
            marker="8", s=550, color="purple", alpha=0.5)
ax2.set_xlabel("Ingreso", fontsize=15)
ax2.set_ylabel("Promedio de Horas Trabajadas", fontsize=15)

plt.show()
```

### Paso 2: Matriz de covarianza

```python
print(datos.cov())

# Código de visualización

plt.figure(figsize=(7, 7))
plt.scatter(datos["ingreso"], datos["horas_trabajadas"], 
            marker="8", s=550, color="purple", alpha=0.5)
plt.xlabel("Ingreso", fontsize=15)
plt.ylabel("Promedio de Horas Trabajadas", fontsize=15)
plt.text(-2.2, 1.5, "Positiva", fontsize=28, color="red")
plt.show()
```

### Paso 3: Eigenvalores y Eigenvectores

```python
from numpy.linalg import eig

# Calcular eigenvalores y eigenvectores de la matriz de covarianza
valores, vectores = eig(datos.cov())

vector_azul = vectores[:, 0]
vector_rojo = vectores[:, 1]

print("Eigenvector Rojo:", vector_rojo, "Eigenvalor:", valores[1])
print("Eigenvector Azul:", vector_azul, "Eigenvalor:", valores[0])

# Código de visualización

plt.figure(figsize=(7, 7))
plt.axes().set_aspect("equal")

# Graficando los datos
plt.scatter(datos["ingreso"], datos["horas_trabajadas"], 
            marker="8", s=550, color="purple", alpha=0.5)

# Graficando los eigenvectores
plt.quiver(0, 0, 
           vector_azul[0]/abs(vector_azul[0])*valores[0],
           vector_azul[1]/abs(vector_azul[1])*valores[0],
           color="blue", angles="xy", scale_units="xy", scale=1, width=0.02)

plt.quiver(0, 0, 
           vector_rojo[0]/abs(vector_rojo[0])*valores[1],
           vector_rojo[1]/abs(vector_rojo[1])*valores[1],
           color="red", angles="xy", scale_units="xy", scale=1, width=0.02)

plt.xlabel("Ingreso", fontsize=15)
plt.ylabel("Promedio de Horas Trabajadas", fontsize=15)
plt.show()
```

### Paso 4: Proyecta datos

```python
proyectados = pd.DataFrame(datos.values @ vectores.T, 
                          columns=["ingreso", "horas_trabajadas"])

# Código de visualización
plt.figure(figsize=(8, 8))
plt.axes().set_aspect("equal")

# Datos rotados
plt.scatter(proyectados["ingreso"], proyectados["horas_trabajadas"], 
            marker="8", s=550, color="purple", alpha=0.5)

# Proyección de los datos en 1 dimensión
plt.scatter(proyectados["ingreso"], [-2]*len(proyectados["ingreso"]), 
            s=250, color="red", alpha=0.5)

plt.scatter([-3]*len(proyectados["horas_trabajadas"]), proyectados["horas_trabajadas"], 
            s=250, color="blue", alpha=0.5)
plt.show()
```

### Paso 5: Obtención de componentes principales

```python
from sklearn.decomposition import PCA

# Obtención de componentes principales
pca = PCA()
datos = pca.fit_transform(datos)

# Código de visualización
plt.figure(figsize=(7, 7))
plt.axes().set_aspect("equal")
plt.scatter(datos[:, 0], [2]*datos[:, 0].size, s=250, color="red", alpha=0.5)
plt.scatter(datos[:, 1], [1]*datos[:, 1].size, s=250, color="blue", alpha=0.5)
plt.ylim((0.8, 2.2))
plt.show()
print("Varianza explicada por cada componente:", pca.explained_variance_)
print("Proporción de varianza explicada por cada componente:", pca.explained_variance_ratio_)
```

### Otro ejemplo

```python
import numpy as np
```

```python
# Creamos una matriz de datos de ejemplo (5 observaciones, 3 variables)
X = np.array([
    [2.5,  2.4,  3.5],
    [0.5,  0.7,  1.2],
    [2.2,  2.9,  3.1],
    [1.9,  2.2,  2.8],
    [3.1,  3.0,  3.6]
])

# Normalizar los datos (media = 0, varianza = 1)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X.shape[1]
```

```python
# Crear el modelo PCA y ajustarlo a los datos
pca = PCA(n_components=X.shape[1] - 1)  # Queremos reducir a 2 dimensiones. Si no se setea nada, por defecto mantiene todos los componentes
pca.fit(X_scaled)
```

```python
print("Componentes principales:")
print(pca.components_)
```

```python
X_pca = pca.transform(X_scaled)
print("Datos transformados:")
print(X_pca)
```

```python
print("Varianza explicada por cada componente:")
print(pca.explained_variance_ratio_)
```

```python
plt.scatter(X_pca[:, 0], X_pca[:, 1], c='blue', label='Datos proyectados')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.title('PCA - Transformación de datos')
plt.legend()
plt.show()
```
